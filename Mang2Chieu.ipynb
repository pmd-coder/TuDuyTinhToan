```python
def matrixx(n):
    full=[]
    matrix=[]
    for i in range(0,n**2):
        full.append(i+1)
    for i in range(0,n**2,n):
        row=full[i:i+n]
        matrix.append(row)
    for i in range(len(matrix)):
        if i%2==1:
            matrix[i].reverse()
    return matrix
n=int(input())
print(matrixx(n))

  

    


```

Bài 3: Trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m. 
Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất. 
Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất. - Dùng vòng lặp để tính tổng từng cột. - Không dùng tuple hoặc dictionary. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
n,m=map(int,input().split())
matrix=[]
tong=0
max=0
index=0
for i in range(n):
    row=list(map(int,input().split()))
    matrix.append(row)
for cot in range(m):
    tong=0
    for hang in range(n):
        tong+=matrix[hang][cot]
        if tong>max:
            max=tong
            index=cot
print(max, index+1)



```
